## 2. LangSmith Datasets

```python
from dotenv import load_dotenv

load_dotenv(override=True, dotenv_path="../.env.local")
```

```python
from langsmith import Client


client = Client()
dataset = client.create_dataset(dataset_name="LangSmith-Apr26-Dataset-2")
import os

# print("API key set:", bool(os.getenv("LANGCHAIN_API_KEY")))
# print("Endpoint:", os.getenv("LANGCHAIN_ENDPOINT"))


examples = [
{"inputs": {"question": "When is John's pay processed?"}, "outputs": {"answer": "John's pay is processed on the 1st of every month."}},
{"inputs": {"question": "What is Julie's job title?"}, "outputs": {"answer": "Julie is a software engineer."}},
]


for ex in examples:
    client.create_example(inputs=ex["inputs"], outputs=ex["outputs"], dataset_id=dataset.id)

print(" Dataset created:", dataset.name)
```
