1. User enters a question
2. programs reads the question

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True, dotenv_path="../.env.local")
my_api_key = os.getenv("OPENAI_API_KEY")

print ( my_api_key[-10:])
```

```python


client = OpenAI()
while (True):
    question = input("What is your question? ")
    print("You asked: " + question)


    if question.lower() == "exit":
        print("Goodbye!")
        break
    else:
        response = client.chat.completions.create(
            model="gpt-4.1-nano", #gpt-5-mini
            messages=[
                {"role": "system", "content": '''
                        You are an information extraction assistant for commercial real estate listings.

                        Your task is to read a block of natural language text provided by the user and extract structured property listing information.

                        The user input may contain descriptions of a commercial property such as location, building size, price, lease information, or agent details.

                        You must extract the information and return it strictly in the following JSON format:

                        {
                        "property_type": "",
                        "address": "",
                        "city": "",
                        "state": "",
                        "zip": "",
                        "units": null,
                        "price": null,
                        "highlights": [],
                        "marketing_blurb": ""
                        }

                        Extraction rules:

                        1. Do NOT hallucinate or invent information.
                        2. Only extract values that are clearly present in the text.
                        3. If a numeric value is unknown, return null.
                        4. If a string value is unknown, return "".
                        5. The highlights field should contain short bullet-style benefits or features extracted from the text.
                        6. The marketing_blurb should be a short 1–2 sentence promotional summary based only on the provided text.

                        Validation rules:

                        If any of the following critical fields are missing:
                        - property_type
                        - city AND state
                        - price

                        Then DO NOT produce the JSON.

                        Instead return 3–7 concise clarifying questions that would help obtain the missing information.

                        Output rules:

                        - Return only JSON OR the list of questions.
                        - Do not include explanations.
                        - Do not include markdown.
                        - Ensure the JSON is valid and properly formatted.
                 
                 '''},
                {"role": "user", "content": question}
            ]
        )
        text_answer = response.choices[0].message.content
        print (text_answer)
```

```python
client = OpenAI()
question = input("What is your question? ")
print("You asked: " + question)


response = client.chat.completions.create(
    model="gpt-4.1-nano", #gpt-5-mini
    messages=[
        {"role": "system", "content": '''You are a helpful assistant.'''},
        {"role": "user", "content": question},

    ],
    temperature=0.7,
    
)
text_answer = response.choices[0].message.content
print (text_answer)
```
