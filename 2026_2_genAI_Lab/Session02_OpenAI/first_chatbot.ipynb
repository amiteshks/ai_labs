## Integration with Open AI

#### 1. Load your OPENAI_API_KEY

```python
import os

from dotenv import load_dotenv

load_dotenv(override=True, dotenv_path="../.env.local")
my_api_key = os.getenv("OPENAI_API_KEY")
```

#### 2. Create a client object

```python
from openai import OpenAI

client = OpenAI(api_key=my_api_key)

client
```

#### 3. Make a call to Open AI - Pass System and User messages


```python
user_input = input("What is your question? ")
prompt = user_input
```

```python
# Multi input model application - Process Input format - Image, Text, Voice etc

# Multi LLM model application - using one model for reasoning (5) and another for answer generation (4.1)

response = client.chat.completions.create(
    model="gpt-4.1-nano", #gpt-5-nano, gpt-5-mini, gpt-5, gpt-5-pro
    messages=[
        {"role": "system", "content": "You are a helpful assistant. Please answer clearly."}, # Single line with double quotes, if we have single line system prompt. If we have multi line system prompt, then we can use triple quotes.
        # {"role": "system", "content": '''You are a helpful assistant. 
        #                                 Please answer clearly.'''},
        {"role": "user", "content": prompt}
    ]
)
```

#### 4. Process the Response & Printing the LLM answer

```python
text_answer_1 = response.choices[0].message.content


print ("-------------  Answer from model 1 (4.1) -------------\n")
print (text_answer_1)
```

```python
with client.audio.speech.with_streaming_response.create(
    model="gpt-4o-mini-tts",
    voice="alloy",
    input=text_answer_1
) as response:
    response.stream_to_file("output.mp3")

print("Audio saved as output.mp3")
```

```python
from IPython.display import Audio
response = client.audio.speech.create(
    model="gpt-4o-mini-tts",
    voice="alloy",
    input=text_answer_1
)

audio_bytes = response.read()

# Play directly in notebook
Audio(audio_bytes)
```
