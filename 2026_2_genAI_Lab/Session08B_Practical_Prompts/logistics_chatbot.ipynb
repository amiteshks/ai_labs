## Integration with Open AI

#### 1. Load your OPENAI_API_KEY

```python
import os

from dotenv import load_dotenv

load_dotenv(override=True, dotenv_path="../.env.local")
my_api_key = os.getenv("OPENAI_API_KEY")
```

#### 2. Create a client object

```python
from openai import OpenAI

client = OpenAI(api_key=my_api_key)

client
```

#### 3. Make a call to Open AI - Pass System and User messages


```python
user_input = input("provide Driver's ID and I will pull up their last shift's data: ")
prompt = user_input
```

```python
# Multi input model application - Process Input format - Image, Text, Voice etc

# Multi LLM model application - using one model for reasoning (5) and another for answer generation (4.1)

response = client.chat.completions.create(
    model="gpt-4.1-nano", #gpt-5-nano, gpt-5-mini, gpt-5, gpt-5-pro
    messages=[
        {"role": "system", "content": '''
            You are an AI assistant that analyzes driver activity data and generates a clear, professional summary of how a driver performed during a shift.
            Return a sample data because we have not connected to the database yet. The data should be in the format mentioned below. 
            Use the data to generate a shift summary, performance rating, key metrics, positives, areas of improvement, recommendations, and risk flags.
            Return Driver ID and Name as well in the output - Use Driver ID as the value provided in the user input and make up a name for the driver.

            =====================================
            INPUT
            =====================================
            You will receive structured and/or semi-structured data about a driver’s shift. This may include:

            - Driver ID / Name
            - Shift start and end time
            - Total driving time
            - Total distance traveled
            - Number of trips / stops
            - Idle time
            - Speeding incidents (count and severity)
            - Harsh braking / acceleration events
            - Route adherence (on-route vs deviations)
            - Fuel or energy usage (if available)
            - Safety score (if available)
            - Any alerts or violations
            - GPS or location patterns (optional)

            =====================================
            OUTPUT REQUIREMENTS
            =====================================

            Return the result in VALID JSON (no markdown, no commentary).

            The output must include:

            {
            "shift_summary": "<concise 3–5 sentence narrative>",
            "performance_rating": "<Excellent | Good | Average | Needs Improvement>",
            "key_metrics": {
                "shift_duration": "",
                "total_distance": "",
                "total_trips": "",
                "idle_time": "",
                "safety_events": "",
                "route_adherence": ""
            },
            "positives": [
                "<list of positive observations>"
            ],
            "areas_of_improvement": [
                "<list of issues or risks>"
            ],
            "recommendations": [
                "<actionable suggestions to improve performance>"
            ],
            "risk_flags": [
                "<any major safety or compliance concerns>"
            ]
            }

            =====================================
            ANALYSIS GUIDELINES
            =====================================

            1. Narrative:
            - Write in a professional, neutral tone
            - Summarize overall driving behavior, efficiency, and safety
            - Highlight both strengths and concerns

            2. Performance Rating:
            - Excellent → Safe, efficient, minimal violations
            - Good → Minor issues but generally solid performance
            - Average → Noticeable inefficiencies or safety events
            - Needs Improvement → Frequent violations, unsafe patterns

            3. Positives:
            - Smooth driving, low idle time, efficient routing, good trip completion

            4. Areas of Improvement:
            - Speeding, harsh driving, excessive idling, route deviation

            5. Recommendations:
            - Be specific and actionable (e.g., “Reduce idle time by shutting off engine during stops >5 mins”)

            6. Risk Flags:
            - Only include serious concerns (e.g., repeated speeding, unsafe driving patterns)

            7. If data is missing:
            - Use reasonable interpretation but do NOT fabricate details

            =====================================
            EXAMPLE STYLE (DO NOT COPY VALUES)
            =====================================

            "shift_summary":
            "The driver completed a full shift with consistent trip activity and moderate efficiency. While route adherence was generally maintained, several speeding incidents and periods of idling were observed. Overall performance indicates acceptable delivery execution but with opportunities to improve safety and fuel efficiency."

            =====================================
            IMPORTANT RULES
            =====================================

            - Always return valid JSON
            - Do NOT include markdown or explanations
            - Do NOT hallucinate metrics
            - Keep insights concise and actionable

        '''}, # Single line with double quotes, if we have single line system prompt. If we have multi line system prompt, then we can use triple quotes.
        # {"role": "system", "content": '''You are a helpful assistant. 
        #                                 Please answer clearly.'''},
        {"role": "user", "content": prompt}
    ]
)
```

#### 4. Process the Response & Printing the LLM answer

```python
text_answer = response.choices[0].message.content
print (text_answer)
```

```python
# Multi input model application - Process Input format - Image, Text, Voice etc

# Multi LLM model application - using one model for reasoning (5) and another for answer generation (4.1)

context = '''John is a software engineer. He has 5 years of experience in Python and JavaScript. 
    He loves to work on open source projects and contribute to the community.'''
prompt = "What is John's profession and how many years of experience does he have in Python and JavaScript?"
response = client.chat.completions.create(
    model="gpt-4.1-nano", #gpt-5-nano, gpt-5-mini, gpt-5, gpt-5-pro
    messages=[
        {"role": "system", "content": "You are a helpful assistant. Please answer clearly only with the information provided in the context."}, # Single line with double quotes, if we have single line system prompt. If we have multi line system prompt, then we can use triple quotes.
        # {"role": "system", "content": '''You are a helpful assistant. 
        #                                 Please answer clearly.'''},
        {"role": "user", "content": f"Context: {context}\n\nQuestion: {prompt}"}
    ]
)
```
