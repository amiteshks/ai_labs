## Integration with Open AI

#### 1. Load your OPENAI_API_KEY

```python
import os

from dotenv import load_dotenv

load_dotenv(override=True, dotenv_path="../.env.local")
my_api_key = os.getenv("OPENAI_API_KEY")
```

#### 2. Create a client object

```python
from openai import OpenAI

client = OpenAI(api_key=my_api_key)

client
```

#### 3. Make a call to Open AI - Pass System and User messages


```python
user_input = input("Enter a calculus test to evaluate: ")
prompt = user_input
```

```python
# Multi input model application - Process Input format - Image, Text, Voice etc

# Multi LLM model application - using one model for reasoning (5) and another for answer generation (4.1)

response = client.chat.completions.create(
    model="gpt-5-nano", #gpt-5-nano, gpt-5-mini, gpt-5, gpt-5-pro
    messages=[
        {"role": "system", "content": '''
            You are an expert calculus teacher and evaluator.

            Your task is to evaluate a student’s answers to a calculus test, assess correctness, identify mistakes, and provide clear, constructive feedback.

            =====================================
            INPUT
            =====================================

            You will receive:

            - A list of questions (with correct answers and/or solution steps if available)
            - The student’s answers (may include steps, final answers, or both)
            - Optional: grading rubric or point allocation

            =====================================
            OUTPUT REQUIREMENTS
            =====================================

            Return VALID JSON only (no markdown, no commentary).

            {
            "summary_section": {
                "overall_score": "<percentage or score>",
                "performance_rating": "<Excellent | Good | Average | Needs Improvement>",
                "overall_feedback": "<2–4 sentence summary of performance>",
                "key_strengths": [
                "<top strengths>"
                ],
                "key_weaknesses": [
                "<top weaknesses>"
                ]
            },
            "detailed_section": {
                "question_evaluations": [
                {
                    "question_id": "",
                    "question_summary": "<short description>",
                    "student_answer": "<as provided>",
                    "correct_answer": "<expected answer>",
                    "is_correct": "<true/false/partial>",
                    "score": "<points awarded>",
                    "max_score": "<total points>",
                    "error_type": "<conceptual | calculation | algebra | incomplete | none>",
                    "feedback": "<clear explanation of what was right/wrong>",
                    "correct_solution_steps": [
                    "<step-by-step correct approach>"
                    ],
                    "improvement_tip": "<how to avoid this mistake in future>"
                }
                ],
                "concept_analysis": {
                "strong_concepts": [
                    "<e.g., derivatives, limits, chain rule>"
                ],
                "weak_concepts": [
                    "<concepts needing improvement>"
                ]
                },
                "common_mistakes": [
                "<patterns observed across answers>"
                ],
                "recommendations": [
                "<actionable study suggestions>"
                ]
            }
            }

            =====================================
            ANALYSIS GUIDELINES
            =====================================

            SCORING:
            - Evaluate each question independently
            - Allow partial credit when steps show correct reasoning
            - Be fair and consistent

            ERROR TYPES:
            - conceptual → misunderstanding of calculus concept
            - calculation → arithmetic mistakes
            - algebra → manipulation errors
            - incomplete → missing steps or final answer

            FEEDBACK STYLE:
            - Be clear, supportive, and instructional
            - Explain WHY something is wrong, not just that it is wrong
            - Highlight correct reasoning even if final answer is wrong

            CORRECT SOLUTIONS:
            - Provide clean, step-by-step solutions
            - Use standard calculus methods (limits, derivatives, integrals, etc.)

            SUMMARY SECTION:
            - Focus on overall performance trends
            - Highlight strengths and improvement areas

            RECOMMENDATIONS:
            - Be specific (e.g., “Practice chain rule problems involving nested functions”)

            MISSING DATA:
            - If correct answers are not provided, infer based on standard calculus knowledge
            - Do NOT fabricate unclear inputs

            =====================================
            IMPORTANT RULES
            =====================================

            - Always return valid JSON
            - No markdown or explanations outside JSON
            - Do not skip any questions
            - Be accurate and educational

        '''}, # Single line with double quotes, if we have single line system prompt. If we have multi line system prompt, then we can use triple quotes.
        # {"role": "system", "content": '''You are a helpful assistant. 
        #                                 Please answer clearly.'''},
        {"role": "user", "content": prompt}
    ]
)
```

#### 4. Process the Response & Printing the LLM answer

```python
text_answer = response.choices[0].message.content
print (text_answer)
```

```python
# Multi input model application - Process Input format - Image, Text, Voice etc

# Multi LLM model application - using one model for reasoning (5) and another for answer generation (4.1)

context = '''John is a software engineer. He has 5 years of experience in Python and JavaScript. 
    He loves to work on open source projects and contribute to the community.'''
prompt = "What is John's profession and how many years of experience does he have in Python and JavaScript?"
response = client.chat.completions.create(
    model="gpt-4.1-nano", #gpt-5-nano, gpt-5-mini, gpt-5, gpt-5-pro
    messages=[
        {"role": "system", "content": "You are a helpful assistant. Please answer clearly only with the information provided in the context."}, # Single line with double quotes, if we have single line system prompt. If we have multi line system prompt, then we can use triple quotes.
        # {"role": "system", "content": '''You are a helpful assistant. 
        #                                 Please answer clearly.'''},
        {"role": "user", "content": f"Context: {context}\n\nQuestion: {prompt}"}
    ]
)
```
