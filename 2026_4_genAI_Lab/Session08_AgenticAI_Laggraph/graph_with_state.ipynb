```python
from typing import TypedDict, Optional
from langgraph.graph import StateGraph, START, END

# class MyState extends TypedDict
class MyState(TypedDict):
    # name: Optional[str]
    # age: Optional[int]
    image_encoding: Optional[str]
    label: str
    order: int
    response: str
    ai_confidence: Optional[float]

# Java
# MyState node_a(MyState state) {... } 
# node_a(initial_state)

def node_a(state: MyState) -> MyState:
    
    print("A ran")
    print(f"state at the beginning of node A: {state}") # state =  {"label": "START NODE", "order": 1, "response": ""}
    
    state["label"] = "Node A"
    state["order"] = state["order"] + 1
    state["response"] = "Hello from Node A"
    
    print(f"state at the end of node A: {state}") # state =  {"label": "START NODE", "order": 1, "response": ""}
    return state # state = {"label": "Node A", "order": 2, "response": "Hello from Node A"}

def node_b(state: MyState) -> MyState:
    print("B ran")
    print(f"state at the beginning of node B: {state}") # state =  {"label": "Node A", "order": 2, "response": "Hello from Node A"}
    
    state["label"] = "Node B"
    state["order"] = state["order"] + 1
    state["response"] = "Hello from Node B"
    image_encoding = state.get("image_encoding", "")
    # call llm with image encoding and get response
    response = call_llm(image_encoding)
    # parse response to get address and update state
    state["response"] = {"address": "123 Main St, Anytown, USA", "price": "$1000000"}
    state["ai_confidence"] = 0.25
    
    print(f"state at the end of node B: {state}") # state =  {"label": "Node B", "order": 3, "response": "Hello from Node B"}
    # print(f"node_b: {state["label"]} is {state["order"]} years old") 
    return state

builder = StateGraph(MyState)
builder.add_node("A", node_a)
builder.add_node("B", node_b)

builder.add_edge(START, "B")
builder.add_edge("B", "A")
builder.add_edge("A", END)

graph = builder.compile()
```
