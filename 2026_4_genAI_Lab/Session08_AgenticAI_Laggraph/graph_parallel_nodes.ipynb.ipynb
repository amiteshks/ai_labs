```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

# class MyState extends TypedDict to define the state structure
class MyState(TypedDict):
    pass

def node_a(state: MyState): 
    print("A ran")
    print("State in A:", state)
    return {}

def node_b(state: MyState): 
    print("B ran")
    print("State in B:", state)
    return {}

def node_c(state: MyState): 
    print("C ran")
    print("State in C:", state)
    return {}

builder = StateGraph(MyState)
builder.add_node("A", node_a)
builder.add_node("B", node_b)
builder.add_node("C", node_c)

builder.add_edge(START, "A")
builder.add_edge("A", "B")
builder.add_edge("A", "C")
builder.add_edge("B", END)
builder.add_edge("C", END)

graph = builder.compile()
print(graph.get_graph().draw_ascii())

state = MyState() # Initialize state as empty dictionary
graph.invoke(state)
```
