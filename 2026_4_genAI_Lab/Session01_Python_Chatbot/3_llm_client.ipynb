```python
import os
from dotenv import load_dotenv

load_dotenv(override=True, dotenv_path="../.env.local")
my_api_key = os.getenv("OPENAI_API_KEY")
# print(my_api_key[-5:])

# print(my_api_key)
```

```python
# %pip install openai
```

```python
from openai import OpenAI
client = OpenAI(api_key=my_api_key)
```

```python
prompt = "Tell me the summary of Arabian Nights."
try:
    response = client.chat.completions.create(
        model="gpt-4.1-nano", #"gpt-5-turbo", #"gpt-5", #"gpt-5-mini", #"gpt-5-nano"
        messages=[
            {"role": "system", "content": "You are a helpful assistant. Answer the user's questions as a JSON object."},
            {"role": "user", "content": prompt}
            # {"role": "user", "content": f"Here is the context: \n\ {context} \n\. \n\n Now, answer the question: {prompt}")
        ],
        # temperature=0.7,
        # max_completion_tokens=256,
    )
    print (response)
    
    print (response.choices[0].message.content)
    
except Exception as e:
    print(f"Error occurred: {e}")

```
