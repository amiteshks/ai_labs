```python
from mlx_lm import load, generate

MODEL = "mlx-community/Llama-3.2-1B-Instruct-4bit"

base_model, tokenizer = load(MODEL)
```

```python
from mlx_lm import load

adapter_model, tokenizer = load(
    MODEL,
    adapter_path="adapters"
)
```

```python
test_prompts = [

"""
Write a listing description for:

22,000 SF Office Building
Palo Alto
Class A
Near Highway 101
""",

"""
Write a professional broker email introducing
a new retail property.
""",

"""
Write an executive summary for an
industrial warehouse in Dallas.
"""
]
```

```python
base_results = []

for prompt in test_prompts:

    response = generate(
        base_model,
        tokenizer,
        prompt=prompt,
        max_tokens=250
    )

    base_results.append(response)
```

```python
finetuned_results = []

for prompt in test_prompts:

    response = generate(
        adapter_model,
        tokenizer,
        prompt=prompt,
        max_tokens=250
    )

    finetuned_results.append(response)
```

```python
for i, prompt in enumerate(test_prompts):

    print("="*80)

    print("PROMPT")
    print(prompt)

    print("\nBASE MODEL\n")
    print(base_results[i])

    print("\nFINE-TUNED MODEL\n")
    print(finetuned_results[i])

    print("="*80)
```
