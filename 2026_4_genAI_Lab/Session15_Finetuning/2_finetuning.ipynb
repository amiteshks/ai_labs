```python
from pathlib import Path

train_file = Path("data/train.jsonl")
validation_file = Path("data/validation.jsonl")

print(train_file.exists())
print(validation_file.exists())
```

```python
import json

count = 0

with open(train_file) as f:
    for line in f:
        json.loads(line)
        count += 1

print("Training Examples:", count)
```

```python
# Parameter	Meaning
# iterations	How long to train
# batch size	Examples per step
# learning rate	How quickly weights change

MODEL = "mlx-community/Llama-3.2-1B-Instruct-4bit"

ITERATIONS = 100

BATCH_SIZE = 2

LEARNING_RATE = 1e-5
```

```python
import subprocess

subprocess.run([
    "python",
    "-m",
    "mlx_lm.lora",
    "--train",
    "--model",
    MODEL,
    "--data",
    "data",
    "--batch-size",
    str(BATCH_SIZE),
    "--iters",
    str(ITERATIONS),
    "--learning-rate",
    str(LEARNING_RATE)
])
```
