#### 2. Create a client object

```python
import ollama


client = ollama.Client()
client
```

#### 3. Make a call to Ollama - Pass System and User messages


```python
user_input = input("What is your question? ")
prompt = user_input
```

```python
# Multi input model application - Process Input format - Image, Text, Voice etc

# Multi LLM model application - using one model for reasoning (5) and another for answer generation (4.1)

response = client.chat(
    model="llama3", #gpt-5-nano, gpt-5-mini, gpt-5, gpt-5-pro
    messages=[
        {"role": "system", "content": "You are a helpful assistant. Please answer clearly."}, # Single line with double quotes, if we have single line system prompt. If we have multi line system prompt, then we can use triple quotes.

        {"role": "user", "content": prompt}
    ]
)
```

#### 4. Process the Response & Printing the LLM answer

```python
print (response.message.content)
```
