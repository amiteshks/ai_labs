```python
#Array
#List
#ArrayList
#Vector

toxic_terms = ["hate", "mean", "stupid"]
text = "I hate this story. It is so stupid and mean."

#Count how many toxic terms are in the text
count = 0
for term in toxic_terms:
    if term in text:
        count += 1

print(count)

if count > 0:
    print("Response can't be generated due to the presence of toxic terms.")
else:    
    print("The text does not contain any toxic terms.")
```
